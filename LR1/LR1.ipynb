# 0. Импортируем необходимые библиотеки

```python
import numpy as np
import pandas as pd
```

# 1. Загружаем датасеты auto1.csv, auto2.csv и помещаем их в датафреймы

```python
auto1 = pd.read_csv('auto1.csv')
auto2 = pd.read_csv('auto2.csv')
```

# 2. В наборе auto1 есть безымянные пустые столбцы, исправим это

```python
auto1 = auto1.loc[:, ~auto1.columns.str.contains('^Unnamed')]
```

# 3. Объединим наборы auto1 и auto2 в один датасет auto

```python
auto = pd.concat([auto1, auto2], ignore_index=True)
```

# 8. Используем группировку и агрегирующие функции

Преобразуем признак horsepower к числовому типу для дальнейших вычислений

```python
auto['horsepower'] = pd.to_numeric(auto['horsepower'], errors='coerce')
```

Определим средний расход топлива, мощность и вес у автомобилей из каждого региона происхождения

```python
stats_by_origin = auto.groupby('origin')[['mpg', 'horsepower', 'weight']].mean()
print('Средние показатели по регионам происхождения:')
print(stats_by_origin)
```

> По результатам группировки по признаку origin видно, что автомобили из региона 3 имеют наибольший средний расход топлива (30.45) и наименьший средний вес (2221.23). Автомобили из региона 1, наоборот, имеют наименьший средний расход топлива (20.08), но при этом наибольшую среднюю мощность (119.05) и наибольший средний вес (3361.93).

Определим, как количество цилиндров влияет на среднюю мощность, ускорение и расход топлива

```python
stats_by_cylinders = auto.groupby('cylinders')[['horsepower', 'acceleration', 'mpg']].mean()
print('Средние показатели в зависимости от количества цилиндров:')
print(stats_by_cylinders)
```

> По результатам группировки по количеству цилиндров видно, что автомобили с большим числом цилиндров в среднем обладают более высокой мощностью и меньшим значением mpg. Наибольшая средняя мощность наблюдается у автомобилей с 8 цилиндрами (158.30), при этом у них же одно из наименьших значений mpg (14.96). Автомобили с 4 цилиндрами, напротив, имеют меньшую среднюю мощность (78.28) и более высокое значение mpg (29.29).

Определим, как менялись средняя мощность двигателей и средний вес автомобилей с течением времени

```python
stats_by_model = auto.groupby('model')[['horsepower', 'weight']].mean()
print('Средняя мощность и средний вес по годам:')
print(stats_by_model)
```

> По результатам группировки по модельному году видно, что в более поздние годы средние значения мощности двигателей и веса автомобилей в целом становятся ниже по сравнению с началом рассматриваемого периода.

# 9. Используем агрегирующие функции agg()

Определим для автомобилей каждого региона средние, минимальные и максимальные значения расхода топлива, медианный вес и стандартное отклонение по мощности

```python
stats_region_agg = auto.groupby('origin').agg({
    'mpg': ['mean', 'min', 'max'],
    'weight': 'median',
    'horsepower': 'std'
})

print('Агрегированные показатели по регионам:')
print(stats_region_agg)
```

> По результатам агрегирования по регионам видно, что автомобили из региона 3 имеют наибольшее среднее значение mpg (30.45), а также максимальное значение mpg, равное 46.6. Автомобили из региона 1 имеют наибольший медианный вес (3365.0) и наибольшее стандартное отклонение мощности (39.90), что говорит о большем разбросе значений мощности в этом регионе.

Определим по каждому году выпуска размах ускорения и суммарную мощность для всех выпущенных в этом году машин

```python
stats_model_agg = auto.groupby('model').agg({
    'acceleration': lambda x: x.max() - x.min(),
    'horsepower': 'sum'
})

print('Размах ускорения и суммарная мощность по годам:')
print(stats_model_agg)
```

> По результатам агрегирования по модельному году видно, что наибольший размах ускорения наблюдается у автомобилей 79 и 82 годов выпуска (13.5 и 13.0 соответственно). Наибольшая суммарная мощность приходится на 73 год выпуска (5219.0), что означает, что в этом году суммарная мощность всех представленных автомобилей была максимальной.

# 10. Используем сводные таблицы pivot_table()

Определим среднюю экономичность автомобилей в зависимости от года выпуска и региона происхождения

```python
pivot_mpg = pd.pivot_table(
    auto,
    values='mpg',
    index='model',
    columns='origin',
    aggfunc='mean'
)

print('Средняя экономичность автомобилей по годам и регионам:')
print(pivot_mpg)
```

> По сводной таблице видно, что в более поздние годы выпуска среднее значение mpg в целом увеличивается для всех регионов происхождения. Наибольшие значения экономичности наблюдаются у автомобилей из регионов 2 и 3 в последние годы рассматриваемого периода.

Создадим сводную таблицу, где для каждого года выпуска и региона происхождения одновременно отображены средний расход топлива и среднее ускорение

```python
pivot_mpg_acc = pd.pivot_table(
    auto,
    values=['mpg', 'acceleration'],
    index='model',
    columns='origin',
    aggfunc='mean'
)

print('Сводная таблица по годам и регионам:')
print(pivot_mpg_acc)
```

> По сводной таблице видно, что для каждого года выпуска и региона происхождения одновременно представлены средние значения расхода топлива и ускорения. Это позволяет сравнивать автомобили разных регионов не только по экономичности, но и по динамическим характеристикам.
